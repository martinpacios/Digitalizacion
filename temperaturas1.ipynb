# Ejercicio para convertir temperaturas Celsius a Fahrenheit

```python
celsius = float(input('Dime la temperatura en ºC: '))
kelvin = celsius + 273.15
print('La temperatura de', celsius,'ºC = ',kelvin,'K')
```
