# CALCULAR PH


```python
ph = float(input("Introduce el valor de ph: "))

if 0 <= ph < 4:
  print ("El valor de ph es muy ácido")
elif 4 <= ph < 7:
  print ("El valor de ph es ácido")
elif ph == 7:
  print ("El valor de ph es neutro")
elif 7 <= ph < 10 :
  print ("El valor de ph es alcalino")
elif 10 <= ph <= 14:
  print ("El valor de ph es muy alcalino")
```
